```python
class Car:
    make = "Toyota" #class/static variable
    def __init__(self, plateNumber, make, model, color, fuel): #constructor
        self.plateNumber = plateNumber
        self.make = make
        self.model = model
        self.color = color
        self.fuel = fuel

    def drive(self, distance):
        fuelUsed = distance * 0.1
        self.fuel =self.fuel - fuelUsed
        if self.fuel < 0:
            self.fuel = 0
            print("Car is out of fuel!")
        print(f"Car drove {distance} km. Remaining fuel: {self.fuel}L")

    def refuel(self, amount):
        self.fuel += amount
        if self.fuel > 100:
            self.fuel = 100
            print("Car is fully refueled!")
        print(f"Car refueled by {amount}L. Current fuel: {self.fuel}L")

    def crash(self, otherCar):
        print(f"{self.plateNumber} crashed into {otherCar.plateNumber}!")
    
    def __repr__(self):
        return f"Car(plateNumber='{self.plateNumber}', make='{self.make}', model='{self.model}', color='{self.color}', fuel={self.fuel})"
```

```python
class Car:
    def __init__(self, plateNumber, make, model, color, fuel): #constructor
        self.plateNumber = plateNumber
        self.make = make
        self.model = model
        self.color = color
        self.fuel = fuel

    def drive(self, distance):
        fuelUsed = distance * 0.1
        self.fuel =self.fuel - fuelUsed
        if self.fuel < 0:
            self.fuel = 0
            print("Car is out of fuel!")
        print(f"Car drove {distance} km. Remaining fuel: {self.fuel}L")

    def refuel(self, amount):
        self.fuel += amount
        if self.fuel > 100:
            self.fuel = 100
            print("Car is fully refueled!")
        print(f"Car refueled by {amount}L. Current fuel: {self.fuel}L")

    def crash(self, otherCar):
        print(f"{self.plateNumber} crashed into {otherCar.plateNumber}!")
        print(self)
        print(otherCar)

    def test(self):
        self.drive()

    def __repr__(self):
        return f"Car(plateNumber='{self.plateNumber}', make='{self.make}', model='{self.model}', color='{self.color}', fuel={self.fuel})"
```

```python
class ElectricCar(Car):
    def __init__(self, plateNumber, make, model, color, fuel, charge):
        Car.__init__(self, plateNumber, make, model, color, fuel)
        self.charge = charge

    def refuel(self,amount):
        self.charge += amount
        if self.charge > 100:
            self.charge = 100
            print("Electric car is fully charged!")
        print(f"Electric car charged by {amount}%. Current charge: {self.charge}%")
    
        
```

```python
myCar = Car("ABC123", "Toyota", "Camry", "Red", 50)
myElectricCar = ElectricCar("ELEC456", "Tesla", "Model 3", "Blue", 0, 80)

myCar.drive(100)
myCar.refuel(30)

myElectricCar.drive(50)
myElectricCar.refuel(15)
```

```python
#interface
class Animal:
    def make_sound(self):
        pass


class Dog(Animal):
    def make_sound(self):
        print("Woof!")

class Cat(Animal):
    def make_sound(self):
        print("Meow!")

class Cow(Animal):
    def make_sound(self):
        print("Moo!")

timmy=Dog()
tily=Cat()
bessie=Cow()

animals=[timmy, tily, bessie]

#interface
def play_sound(animal): #doesnt care about the type of animal, just needs to know that it can make a sound
    animal.make_sound()


for animal in animals:
    play_sound(animal)
```

```python
class Student:
    def __init__(self, nic,name, age, registered_course):
        self.__nic= nic #private variable
        self.name = name
        self.age = age
        self.registered_course = registered_course

    def change_nic(self, new_nic):
        self.__nic = new_nic
   

    def __repr__(self):
        return f"Student(nic='{self.__nic}', name='{self.name}', age={self.age}, registered_course='{self.registered_course}')"

#main program
tomy=Student("123456789V", "Tommy", 20, "Data Science")
print(tomy)

tomy.__nic="987654321V" #this will not change the nic because it is private
print(tomy)

tomy.change_nic("987654321V") #this will change the nic because we are using the public method
print(tomy)
```
