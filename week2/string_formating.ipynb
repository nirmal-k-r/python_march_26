```python
name="Paul"
age=21
married=False

print(name)
print(age)
print(married)
```

```python
print(name, age, married)
```

```python
print(name, "is", age, "years old. Married:", married)
```

```python
#string formating
name="Paul"
age=21
married=False

print("{} is {} years old. Married: {}".format(name, age, married))
```

```python
#string interpolation - f string
print(f'My name is {name}')

print(f'{name} is {age} years old. Married: {married}')
```
