```python
#write a program to calculate fibinacci sequence 
# ex:  the 1 2 3 5 8

num1=1
num2=2


print(num1)
print(num2)

for i in range(50):
    num3=num1+num2
    num1=num2
    num2=num3
    print(num3)
```
