```python
print('Program started')
try:
    x=10/1
    print("Result:",x)
except:
    print("Error occured")
else: #only runs if no exception occurs
    print("No error occured")
finally: #code block which runs regardless of an exception
    print("This will always run")
     
print('Program ended')
```
