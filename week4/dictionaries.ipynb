```python
#accessing dictionary values
marks={
    "english": 90,
    "maths": 95,
    "science": 85,
    "social": 80
}

marks["maths"]
marks["english"]
```

```python
#modifying dictionary values

marks={
    "english": 90,
    "maths": 95,
    "science": 85,
    "social": 80
}

marks["social"]=100

marks["english"]=99

print(marks)
```

```python
#appending to the dictionary
data={}

data["country"]="Mauritius"

print(data)

data["population"]=1000000
print(data)
```

```python
#delete from the dictionary

del data["population"]
print(data)
```

```python
marks={
    "english": 90,
    "maths": 95,
    "science": 85,
    "social": 80
}


# list(marks.keys())[1]
print(marks.keys())
print(marks.values())
```

```python
marks={
    "english": 90,
    "maths": 95,
    "science": 85,
    "social": 80
}

for key in marks.keys(): #for key in marks:
    print(key)
```

```python
for key in marks.values(): #for key in marks:
    print(key)
```

```python
print(marks.items())
```

```python
#iterate over all keys and values in a dictionary
for key,value in marks.items():
    print(key,value)
```
